```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_excel('data.xlsx')
data.head()
```

```python
%config InlineBackend.figure_format = 'svg'
```

```python
sns.set_context("paper")
sns.set_style("ticks")

plt.rcParams.update({
    "font.size": 8,
    "axes.labelsize": 10,
    "figure.titlesize": 10,
    "legend.fontsize": 8,
    "xtick.direction": "in",
    "ytick.direction": "in"
})

fig, ax = plt.subplots(figsize=(3.5, 3.0))

sns.scatterplot(
    data=data,
    x='CO2 Density (kg/m3)',
    y='P (MPa)',
    hue='T (°C)',
    style='Secure',
    edgecolors='black',
    linewidth=0.5,
    alpha=0.8,
    s=40,
    palette=sns.diverging_palette(145, 300, s=60, as_cmap=True),
    ax=ax
);

sns.despine(ax=ax)
ax.set_xlim(0, 1000)
ax.set_ylim(0, 80);

ax.legend(
    loc="upper left",
    bbox_to_anchor=(1, 1),
    frameon=False
);

ax.set_xlabel('CO$_2$ Density (kg/m$^3$)')
ax.set_ylabel('Pressure (MPa)');

# plt.savefig(
#     "co2_phase_plot.pdf",
#     bbox_inches="tight"
# );

plt.savefig(
    "co2_phase_plot.png",
    bbox_inches="tight",
    dpi=300,
    transparent=True
);
```
